# Smartphone Addiction – Random Forest

Same data prep and same train/test split as the neural-network notebook, so results are directly comparable.
Tasks: (A) 4-class `addiction_level` (None / Mild / Moderate / Severe), (B) binary addicted vs. not addicted.

```python
import warnings; warnings.filterwarnings("ignore")
import os, glob
import numpy as np, pandas as pd, matplotlib.pyplot as plt, joblib
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.inspection import permutation_importance
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
SEED = 42
%matplotlib inline
```

```python
CSV = "Smartphone_Usage_And_Addiction_Analysis_7500_Rows.csv"   # <- change if needed
if not os.path.exists(CSV):
    hits = glob.glob("**/*Smartphone_Usage_And_Addiction_Analysis_7500_Rows.csv", recursive=True) \
         + glob.glob("/mnt/user-data/uploads/*Smartphone_Usage_And_Addiction_Analysis_7500_Rows.csv")
    CSV = hits[0]
df = pd.read_csv(CSV)
df["addiction_level"] = df["addiction_level"].fillna("None")

order = ["None", "Mild", "Moderate", "Severe"]
X = df.drop(columns=["transaction_id", "user_id", "addicted_label", "addiction_level"])
y = df["addiction_level"].map({c: i for i, c in enumerate(order)})
yb = df["addicted_label"]

Xd = pd.get_dummies(X, drop_first=True)          # trees don't need scaling
X_tr, X_te, y_tr, y_te = train_test_split(Xd, y, test_size=0.2, stratify=y, random_state=SEED)
print(Xd.shape)
```

## A. Four-class random forest (tuned with 5-fold CV)

```python
grid = {
    "n_estimators": [300],
    "max_depth": [8, None],
    "min_samples_leaf": [5, 10, 20],
    "max_features": ["sqrt"],
}
cv = StratifiedKFold(5, shuffle=True, random_state=SEED)
gs = GridSearchCV(RandomForestClassifier(random_state=SEED, n_jobs=-1), grid,
                  cv=cv, scoring="accuracy", n_jobs=-1).fit(X_tr, y_tr)
print("Best params:", gs.best_params_)
print(f"Best CV accuracy: {gs.best_score_:.4f}")
```

```python
rf = gs.best_estimator_
pred = rf.predict(X_te)
print(f"Test accuracy: {accuracy_score(y_te, pred):.4f}\n")
print(classification_report(y_te, pred, target_names=order, digits=3))
ConfusionMatrixDisplay(confusion_matrix(y_te, pred), display_labels=order).plot(cmap="Blues", colorbar=False)
plt.title("Random forest – confusion matrix (test)"); plt.show()
```

```python
imp = permutation_importance(rf, X_te, y_te, n_repeats=10, random_state=SEED, n_jobs=-1)
pd.Series(imp.importances_mean, index=Xd.columns).sort_values().plot.barh(figsize=(7,5), color="tab:green")
plt.xlabel("Drop in accuracy when shuffled"); plt.title("Random forest – permutation importance"); plt.tight_layout(); plt.show()
```

## B. Binary random forest: addicted vs. not addicted

```python
Xb_tr, Xb_te, yb_tr, yb_te = train_test_split(Xd, yb, test_size=0.2, stratify=yb, random_state=SEED)
gsb = GridSearchCV(RandomForestClassifier(random_state=SEED, n_jobs=-1),
                   {"n_estimators": [300], "max_depth": [8, None], "min_samples_leaf": [5, 10, 20]},
                   cv=cv, scoring="accuracy", n_jobs=-1).fit(Xb_tr, yb_tr)
rfb = gsb.best_estimator_
pb = rfb.predict(Xb_te)
print("Best params:", gsb.best_params_)
print(f"Binary test accuracy: {accuracy_score(yb_te, pb):.4f}  (majority baseline {yb.mean():.3f})\n")
print(classification_report(yb_te, pb, target_names=["Not addicted", "Addicted"], digits=3))
```

## C. Random forest vs. neural network (same splits)

```python
cat_cols = ["gender", "stress_level", "academic_work_impact"]
num_cols = [c for c in X.columns if c not in cat_cols]
pre = ColumnTransformer([("num", StandardScaler(), num_cols), ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols)])
def mlp(): return Pipeline([("prep", pre), ("mlp", MLPClassifier((64, 32), alpha=1e-3, batch_size=64,
                 early_stopping=True, max_iter=300, n_iter_no_change=15, random_state=SEED))])

Xr_tr, Xr_te = X.loc[X_tr.index], X.loc[X_te.index]
nn4 = mlp().fit(Xr_tr, y_tr)
nn2 = mlp().fit(X.loc[Xb_tr.index], yb_tr)

table = pd.DataFrame({
    "4-class accuracy":  [accuracy_score(y_te, nn4.predict(Xr_te)), accuracy_score(y_te, pred)],
    "Binary accuracy":   [accuracy_score(yb_te, nn2.predict(X.loc[Xb_te.index])), accuracy_score(yb_te, pb)],
}, index=["Neural network (MLP)", "Random forest"]).round(4)
table
```

```python
joblib.dump({"model": rf, "classes": order, "columns": list(Xd.columns)}, "rf_addiction_4class.joblib")
joblib.dump({"model": rfb, "columns": list(Xd.columns)}, "rf_addiction_binary.joblib")
```
