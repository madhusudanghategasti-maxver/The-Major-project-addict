# Smartphone Addiction Prediction — Model Comparison
Dataset: [Kaggle – smartphone-addiction-prediction](https://www.kaggle.com/datasets/danishzulfiqar5050/smartphone-addiction-prediction)

**Plan:** load data → EDA → preprocess → compare Logistic Regression / Random Forest / Gradient Boosting / XGBoost with cross-validation → tune best → evaluate on test set → feature importance.

```python
# !pip install pandas scikit-learn matplotlib seaborn xgboost
import glob, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (ConfusionMatrixDisplay, RocCurveDisplay,
                             classification_report, roc_auc_score)
from sklearn.model_selection import (GridSearchCV, StratifiedKFold,
                                     cross_validate, train_test_split)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

try:
    from xgboost import XGBClassifier
    HAS_XGB = True
except ImportError:
    HAS_XGB = False
    print("xgboost not installed - skipping it")
```

## 1. Load data

```python
# Set this to your CSV path. On Kaggle it auto-detects the file under /kaggle/input.
CSV_PATH = None
if CSV_PATH is None:
    found = glob.glob("/kaggle/input/**/*.csv", recursive=True) or glob.glob("*.csv")
    CSV_PATH = found[0]
print("Using:", CSV_PATH)

df = pd.read_csv(CSV_PATH)
print(df.shape)
df.head()
```

```python
df.info()
df.describe(include="all").T
```

## 2. Choose target column
Defaults to the last column. Change `TARGET` if the label is elsewhere.

```python
TARGET = df.columns[-1]   # <- change if needed
print("Target:", TARGET)

# Drop ID-like columns and rows missing the target
id_like = [c for c in df.columns if c != TARGET and (c.lower() == "id" or c.lower().endswith("_id"))]
df = df.drop(columns=id_like).dropna(subset=[TARGET])

X = df.drop(columns=[TARGET])
y_raw = df[TARGET]
if y_raw.dtype == object or y_raw.dtype == bool or y_raw.nunique() <= 10:
    le = LabelEncoder()
    y = pd.Series(le.fit_transform(y_raw), index=y_raw.index)
    print("Classes:", dict(zip(le.classes_, le.transform(le.classes_))))
else:
    raise ValueError("Target looks continuous - threshold it into classes or pick another column.")

multiclass = y.nunique() > 2
y.value_counts(normalize=True).round(3)
```

## 3. Quick EDA

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
y.value_counts().plot(kind="bar", ax=axes[0], title="Class distribution")
num_cols_all = X.select_dtypes(include=np.number).columns
if len(num_cols_all) > 1:
    sns.heatmap(X[num_cols_all].corr(), cmap="coolwarm", center=0, ax=axes[1])
    axes[1].set_title("Numeric feature correlations")
plt.tight_layout(); plt.show()
```

## 4. Preprocessing & split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

num_cols = X_train.select_dtypes(include=np.number).columns.tolist()
cat_cols = [c for c in X_train.columns if c not in num_cols]

preprocessor = ColumnTransformer([
    ("num", Pipeline([("imp", SimpleImputer(strategy="median")),
                      ("sc", StandardScaler())]), num_cols),
    ("cat", Pipeline([("imp", SimpleImputer(strategy="most_frequent")),
                      ("oh", OneHotEncoder(handle_unknown="ignore"))]), cat_cols),
])
print(f"{len(num_cols)} numeric, {len(cat_cols)} categorical features")
```

## 5. Compare models (5-fold stratified CV)

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=2000, class_weight="balanced"),
    "Random Forest": RandomForestClassifier(n_estimators=300, class_weight="balanced",
                                            random_state=42, n_jobs=-1),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
}
if HAS_XGB:
    models["XGBoost"] = XGBClassifier(n_estimators=300, learning_rate=0.05, max_depth=4,
                                      eval_metric="logloss", random_state=42)

scoring = {"accuracy": "accuracy",
           "f1": "f1_macro" if multiclass else "f1",
           "roc_auc": "roc_auc_ovr" if multiclass else "roc_auc"}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
rows = {}
for name, model in models.items():
    pipe = Pipeline([("pre", preprocessor), ("clf", model)])
    res = cross_validate(pipe, X_train, y_train, cv=cv, scoring=scoring)
    rows[name] = {m: res[f"test_{m}"].mean() for m in scoring}

results = pd.DataFrame(rows).T.sort_values("f1", ascending=False)
results.round(4)
```

```python
results.plot(kind="bar", figsize=(10, 4), ylim=(0, 1.05), title="CV performance by model")
plt.xticks(rotation=20); plt.tight_layout(); plt.show()
best_name = results.index[0]
print("Best model by CV F1:", best_name)
```

## 6. Tune the best model

```python
grids = {
    "Random Forest": {"clf__max_depth": [None, 5, 10], "clf__min_samples_leaf": [1, 3, 5]},
    "Gradient Boosting": {"clf__learning_rate": [0.05, 0.1], "clf__max_depth": [2, 3, 4]},
    "XGBoost": {"clf__max_depth": [3, 4, 6], "clf__learning_rate": [0.03, 0.05, 0.1]},
    "Logistic Regression": {"clf__C": [0.01, 0.1, 1, 10]},
}
pipe = Pipeline([("pre", preprocessor), ("clf", models[best_name])])
search = GridSearchCV(pipe, grids[best_name], cv=cv, scoring=scoring["f1"], n_jobs=-1)
search.fit(X_train, y_train)
print("Best params:", search.best_params_)
print("Best CV F1:", round(search.best_score_, 4))
best_model = search.best_estimator_
```

## 7. Final evaluation on held-out test set

```python
preds = best_model.predict(X_test)
proba = best_model.predict_proba(X_test)
print(classification_report(y_test, preds))
auc = (roc_auc_score(y_test, proba, multi_class="ovr") if multiclass
       else roc_auc_score(y_test, proba[:, 1]))
print("Test ROC-AUC:", round(auc, 4))

fig, axes = plt.subplots(1, 2 if not multiclass else 1, figsize=(11, 4))
axes = np.atleast_1d(axes)
ConfusionMatrixDisplay.from_predictions(y_test, preds, ax=axes[0])
axes[0].set_title("Confusion matrix")
if not multiclass:
    RocCurveDisplay.from_predictions(y_test, proba[:, 1], ax=axes[1])
    axes[1].set_title("ROC curve")
plt.tight_layout(); plt.show()
```

## 8. Feature importance

```python
clf = best_model.named_steps["clf"]
names = best_model.named_steps["pre"].get_feature_names_out()

if hasattr(clf, "feature_importances_"):
    imp = pd.Series(clf.feature_importances_, index=names)
elif hasattr(clf, "coef_"):
    imp = pd.Series(np.abs(clf.coef_).mean(axis=0), index=names)
else:
    imp = None

if imp is not None:
    imp.sort_values().tail(15).plot(kind="barh", figsize=(8, 6), title="Top 15 features")
    plt.tight_layout(); plt.show()
```
