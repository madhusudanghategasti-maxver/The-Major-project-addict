# Smartphone Addiction Prediction — Random Forest
Predict `addicted_label` (1 = addicted, 0 = not addicted) from smartphone usage behaviour.

**Note:** update `DATA` below if your CSV is in a different location.

```python
import pandas as pd, numpy as np, joblib
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import (classification_report, confusion_matrix, ConfusionMatrixDisplay,
                             roc_auc_score, RocCurveDisplay, accuracy_score)

DATA = "Smartphone_Usage_And_Addiction_Analysis_7500_Rows.csv"
df = pd.read_csv(DATA)
print(df.shape)
df.head()
```

## 1. Explore the data

```python
print(df.isna().sum())
print(df.addicted_label.value_counts(normalize=True))
df.describe().T
```

## 2. Target & leakage
`addiction_level` (Mild/Moderate/Severe/NaN) is a finer version of the target — Moderate/Severe always map to 1, Mild/NaN to 0 — so it **must be excluded** or the model would simply read the answer. ID columns are dropped too.

```python
print(pd.crosstab(df.addiction_level.fillna('None'), df.addicted_label))
```

```python
y = df["addicted_label"]
X = df.drop(columns=["addicted_label", "addiction_level", "transaction_id", "user_id"])

# Feature engineering
X["weekend_minus_weekday"] = X["weekend_screen_time"] - X["daily_screen_time_hours"]
X["opens_per_notification"] = X["app_opens_per_day"] / (X["notifications_per_day"] + 1)
X["social_share"] = X["social_media_hours"] / X["daily_screen_time_hours"]

cat_nominal = ["gender", "academic_work_impact"]
cat_ordinal = ["stress_level"]
num_cols = [c for c in X.columns if c not in cat_nominal + cat_ordinal]

pre = ColumnTransformer([
    ("nom", OneHotEncoder(drop="if_binary", handle_unknown="ignore"), cat_nominal),
    ("ord", OrdinalEncoder(categories=[["Low", "Medium", "High"]]), cat_ordinal),
    ("num", "passthrough", num_cols),
])

pipe = Pipeline([("pre", pre),
                 ("rf", RandomForestClassifier(random_state=42, n_jobs=-1, class_weight="balanced"))])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
X_train.shape, X_test.shape
```

## 3. Hyperparameter tuning
Small grid to keep runtime short — widen it (e.g. more `n_estimators`, `max_features=0.5`) if you have more CPU cores.

```python
grid = {
    "rf__n_estimators": [200],
    "rf__max_depth": [None, 10],
    "rf__min_samples_leaf": [1, 5],
    "rf__max_features": ["sqrt"],
}
cv = StratifiedKFold(5, shuffle=True, random_state=42)
search = GridSearchCV(pipe, grid, cv=3, scoring="roc_auc", n_jobs=-1)
search.fit(X_train, y_train)
best = search.best_estimator_
print("Best params:", search.best_params_)
print(f"CV ROC-AUC (best): {search.best_score_:.4f}")
```

## 4. Evaluation

```python
pred = best.predict(X_test)
proba = best.predict_proba(X_test)[:, 1]
print(f"Test accuracy: {accuracy_score(y_test, pred):.4f}")
print(f"Test ROC-AUC : {roc_auc_score(y_test, proba):.4f}\n")
print(classification_report(y_test, pred, target_names=["Not addicted", "Addicted"]))
print("5-fold accuracy:", cross_val_score(best, X, y, cv=cv, scoring="accuracy").round(4))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))
ConfusionMatrixDisplay.from_predictions(y_test, pred, display_labels=["Not addicted", "Addicted"], ax=ax[0], cmap="Blues")
ax[0].set_title("Confusion Matrix")
RocCurveDisplay.from_predictions(y_test, proba, ax=ax[1]); ax[1].set_title("ROC Curve")
plt.tight_layout(); plt.show()
```

## 5. Feature importance

```python
names = [n.split("__")[-1] for n in best.named_steps["pre"].get_feature_names_out()]
imp = pd.Series(best.named_steps["rf"].feature_importances_, index=names).sort_values()
imp.plot.barh(figsize=(8, 6), title="Feature Importance")
plt.tight_layout(); plt.show()
imp.sort_values(ascending=False).round(4)
```

## 6. Save the model & predict on new data

```python
joblib.dump(best, "random_forest_addiction_model.joblib")

# Example: predict for a new user
new_user = pd.DataFrame([{
    "age": 24, "gender": "Female", "daily_screen_time_hours": 9.5, "social_media_hours": 4.5,
    "gaming_hours": 1.5, "work_study_hours": 3.0, "sleep_hours": 6.0,
    "notifications_per_day": 180, "app_opens_per_day": 120, "weekend_screen_time": 11.5,
    "stress_level": "High", "academic_work_impact": "Yes"}])
new_user["weekend_minus_weekday"] = new_user.weekend_screen_time - new_user.daily_screen_time_hours
new_user["opens_per_notification"] = new_user.app_opens_per_day / (new_user.notifications_per_day + 1)
new_user["social_share"] = new_user.social_media_hours / new_user.daily_screen_time_hours
print("Prediction:", best.predict(new_user)[0], "| P(addicted) =", round(best.predict_proba(new_user)[0, 1], 3))
```
